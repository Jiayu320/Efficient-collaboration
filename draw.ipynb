```python
# 首先需要安装 plotly
# pip install plotly pandas

import pandas as pd
import plotly.express as px
import io

# 1. 准备数据 (和之前一样)
csv_data = """Planner Model,Plan Soundness,Dependency Structure,Task Clarity,Attribute Accuracy,Plan Efficiency
gpt-5,5.00,4.80,4.64,3.84,4.84
gemini-2.5-pro,4.87,4.93,5.00,3.90,4.87
llama-3-8b-instruct,2.37,3.60,3.73,2.57,2.53
"""
df = pd.read_csv(io.StringIO(csv_data))

# Plotly 需要"长格式"数据，我们用 melt() 函数轻松转换
df_long = pd.melt(df, id_vars=['Planner Model'], var_name='Metric', value_name='Score')

# 2. 一行核心代码绘图！
fig = px.line_polar(df_long, 
                    r='Score',               # r 是半径，代表分数
                    theta='Metric',            # theta 是角度，代表评估维度
                    color='Planner Model',     # color 用来区分不同模型
                    line_close=True,           # line_close=True 自动闭合线条
                    range_r=[0, 5.5],          # 设置半径范围
                    title='Planner Model Performance Comparison',
                    template='plotly_white'     # 可以试试不同的主题模板, e.g., 'plotly_white', 'ggplot2', 'seaborn'
                   )

# 3. 更新布局，让图表更好看
fig.update_traces(fill='toself', fillcolor=None, opacity=0.8) # 填充区域
fig.update_layout(
    font_size=14,
    legend_font_size=16,
    title_font_size=24
)


fig.show() # 在浏览器或 Notebook 中显示交互式图表
```
