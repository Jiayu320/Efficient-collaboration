```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
import io

# 1. 准备数据 (与您提供的一致)
csv_data = """Planner Model,Plan Soundness,Dependency Structure,Task Clarity,Attribute Accuracy,Plan Efficiency
qwen3-235b-a22b-thinking-2507,4.86,4.86,4.86,3.66,4.86
gemini-2.5-pro,4.87,4.93,5.00,3.90,4.87
llama-3-8b-instruct,2.37,3.60,3.73,2.57,2.53
our-model(4b), 4.20,4.60,4.60,3.20,4.40
our-model(1.7b), 2.60,3.80,4.60,2.40,2.60
"""

df = pd.read_csv(io.StringIO(csv_data))

# 将数据转换为 Plotly 喜欢的 "长格式"
df_long = pd.melt(df, id_vars=['Planner Model'], var_name='Metric', value_name='Score')

# 2. 定义样式 - 在这里突出您自己的模型
color_discrete_map = {
    'our-model(4b)': 'crimson',          # 保持亮色突出
    'our-model(1.7b)': 'deepskyblue',      # 保持亮色突出
    'qwen3-235b-a22b-thinking-2507': 'slategray',        # 新颜色1: 石板灰
    'gemini-2.5-pro': 'darkseagreen',    # 新颜色2: 暗橄榄绿
    'llama-3-8b-instruct': 'sienna'          # 新颜色3: 赭色
}

line_dash_map = {
    'our-model(4b)': 'solid',
    'our-model(1.7b)': 'solid',
    'qwen3-235b-a22b-thinking-2507': 'dash',
    'gemini-2.5-pro': 'dash',
    'llama-3-8b-instruct': 'dashdot'
}


# 3. 使用 Plotly Express 创建基础雷达图
fig = px.line_polar(df_long,
                    r='Score',
                    theta='Metric',
                    color='Planner Model',
                    line_close=True,
                    range_r=[0, 5.5],
                    title='Planner Model Performance Comparison',
                    template='plotly_white',
                    color_discrete_map=color_discrete_map,
                    line_dash='Planner Model',
                    line_dash_map=line_dash_map
                   )

# 4. 更新线条细节
fig.update_traces(fill='none', line_width=2)

for trace in fig.data:
    if 'our-model' in trace.name:
        trace.update(line_width=3)

# 5. 更新布局 (这是修复了 polar 内部错误的最终版本)
fig.update_layout(
    font=dict(
        size=15
    ),
    title=dict(
        font=dict(
            size=24
        )
    ),
    legend=dict(
        title_text='',
        font=dict(
            size=16
        )
    ),
    # 设置极坐标轴（雷达图的轴）
    polar=dict(
        radialaxis=dict(
            visible=True,
            range=[0, 5.5],
            angle=90,
            # *** 此处已修正: font -> tickfont ***
            tickfont=dict(size=12)
        ),
        angularaxis=dict(
          # *** 此处已修正: font -> tickfont ***
          tickfont=dict(size=16)
        )
    )
)

fig.show()
```
